```python

import random

board = [' ' for i in range(9)]


def show_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner(symbol):
    win_positions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for a, b, c in win_positions:
        if board[a] == board[b] == board[c] == symbol:
            return True

    return False


def is_full():
    return ' ' not in board


def computer_move():
    empty = [i for i in range(9) if board[i] == ' ']
    choice = random.choice(empty)
    board[choice] = 'O'
    print("Computer selected position:", choice + 1)


print("TIC-TAC-TOE GAME")
print("Player: X")
print("Computer: O")

print("\nBoard positions:")
print("1 | 2 | 3")
print("--+---+--")
print("4 | 5 | 6")
print("--+---+--")
print("7 | 8 | 9")

while True:
    show_board()

    while True:
        try:
            pos = int(input("Choose a position (1-9): "))

            if pos < 1 or pos > 9:
                print("Please enter a number from 1 to 9.")
            elif board[pos - 1] != ' ':
                print("That position is occupied. Try again.")
            else:
                board[pos - 1] = 'X'
                break

        except ValueError:
            print("Invalid input. Enter a number.")

    if check_winner('X'):
        show_board()
        print("Player wins!")
        break

    if is_full():
        show_board()
        print("It's a draw!")
        break

    computer_move()

    if check_winner('O'):
        show_board()
        print("Computer wins!")
        break

    if is_full():
        show_board()
        print("It's a draw!")
        break
```

```python

room_A = input("Enter Room A status (Dirty/Clean): ").capitalize()
room_B = input("Enter Room B status (Dirty/Clean): ").capitalize()

location = input("Enter starting location (A/B): ").upper()

rooms = {
    'A': room_A,
    'B': room_B
}

print("\nInitial Room Status:")
print(rooms)
print("Starting Location:", location)

while True:
    if rooms[location] == "Dirty":
        print("Action: Clean")
        rooms[location] = "Clean"

    else:
        print("Action: MOVE")

        if location == 'A':
            location = 'B'
        else:
            location = 'A'

    if rooms['A'] == "Clean" and rooms['B'] == "Clean":
        break

print("\nFinal Room Status:")
print(rooms)
print("Final Location:", location)
print("Both rooms are clean!")
```
