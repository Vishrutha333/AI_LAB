```python

import heapq

def heuristic(state, goal):
    return sum(1 for i in range(9)
               if state[i] != 0 and state[i] != goal[i])

def astar(start, goal):
    pq = [(heuristic(start, goal), 0, start)]
    parent = {start: None}
    cost = {start: 0}

    while pq:
        f, g, current = heapq.heappop(pq)

        if g != cost[current]:
            continue

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            return path[::-1]

        zero = current.index(0)
        row, col = divmod(zero, 3)

        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                pos = nr * 3 + nc
                new_state = list(current)
                new_state[zero], new_state[pos] = new_state[pos], new_state[zero]
                neighbor = tuple(new_state)
                new_g = g + 1

                if new_g < cost.get(neighbor, float("inf")):
                    cost[neighbor] = new_g
                    parent[neighbor] = current
                    heapq.heappush(
                        pq, (new_g + heuristic(neighbor, goal),
                             new_g, neighbor)
                    )

    return None

def display(path):
    if path is None:
        print("No solution found")
        return

    print("Number of moves:", len(path) - 1)
    for step, state in enumerate(path):
        print("\nStep", step)
        for i in range(0, 9, 3):
            print(state[i:i + 3])

start = tuple(map(int, input("Enter initial state (0 for blank): ").split()))
goal = tuple(map(int, input("Enter goal state (0 for blank): ").split()))

if (len(start) != 9 or len(goal) != 9
        or set(start) != set(range(9))
        or set(goal) != set(range(9))):
    print("Invalid input! Enter numbers 0 to 8 exactly once.")
else:
    display(astar(start, goal))
```
