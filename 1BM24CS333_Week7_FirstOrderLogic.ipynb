```python

def is_variable(x):
    return isinstance(x, str) and len(x) == 1 and x.islower()

def substitute(x, subs):
    while isinstance(x, str) and x in subs:
        x = subs[x]
    if isinstance(x, tuple):
        return tuple(substitute(a, subs) for a in x)
    return x

def occurs(var, expr, subs):
    expr = substitute(expr, subs)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs(var, a, subs) for a in expr)

    return False

def unify(x, y, subs):
    x = substitute(x, subs)
    y = substitute(y, subs)

    if x == y:
        return subs

    if is_variable(x):
        if occurs(x, y, subs):
            return None
        subs[x] = y
        return subs

    if is_variable(y):
        if occurs(y, x, subs):
            return None
        subs[y] = x
        return subs

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subs = unify(a, b, subs)
            if subs is None:
                return None

        return subs

    return None

def get_expression():
    s = input().strip()
    name, args = s.split("(")
    args = args.rstrip(")").split(",")
    return (name.strip(),) + tuple(a.strip() for a in args)

while True:
    print("\n--- UNIFICATION ---")

    print("Enter first expression (e.g. eats(x,apple)):")
    E1 = get_expression()

    print("Enter second expression (e.g. eats(riya,y)):")
    E2 = get_expression()

    result = unify(E1, E2, {})

    if result is None:
        print("\nUnification Failed")
    else:
        print("\nUnification Successful")
        print("Substitution:", result)
        print("Unified Expression:", substitute(E1, result))

    choice = input("\nDo you want to continue? (yes/no): ").lower()

    if choice != "yes":
        print("Program Exited.")
        break
```
