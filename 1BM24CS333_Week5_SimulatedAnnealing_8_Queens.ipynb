```python

import random
import math

N = 8

def calculate_cost(state):
    conflicts = 0
    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts


def get_neighbor(state):
    neighbor = state.copy()
    col = random.randrange(N)
    row = random.randrange(N - 1)

    if row >= neighbor[col]:
        row += 1

    neighbor[col] = row
    return neighbor


def simulated_annealing(initial):
    current = initial.copy()
    cost = calculate_cost(current)
    temperature = 100.0
    step = 0

    print("Initial:", current, "Cost:", cost)

    while cost > 0 and temperature > 0.01:
        neighbor = get_neighbor(current)
        new_cost = calculate_cost(neighbor)
        delta = new_cost - cost

        if delta < 0 or random.random() < math.exp(-delta / temperature):
            current = neighbor
            cost = new_cost
            step += 1
            print("Step", step, ":", current, "Cost:", cost)

        temperature *= 0.95

    return current, cost


initial = list(map(int, input("Enter 8 row values (0-7): ").split()))

if len(initial) != 8 or any(x < 0 or x >= 8 for x in initial):
    print("Invalid input")
else:
    solution = None

    for attempt in range(100):
        start = initial if attempt == 0 else [random.randrange(8) for _ in range(8)]
        solution, cost = simulated_annealing(start)

        if cost == 0:
            print("\nSolution found!")
            print("Final configuration:", solution)
            print("Final cost:", cost)
            break
    else:
        print("Try running again.")
```
