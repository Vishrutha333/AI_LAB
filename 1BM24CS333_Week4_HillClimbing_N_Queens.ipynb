```python

import random

N = 4

def calculate_cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def display_board(state):
    print("Current board:")
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


def get_neighbors(state):
    neighbors = []

    for col in range(N):
        for row in range(N):
            if row != state[col]:
                neighbor = state.copy()
                neighbor[col] = row
                neighbors.append(neighbor)

    return neighbors


def hill_climbing(state):
    current = state.copy()

    while True:
        current_cost = calculate_cost(current)

        print("\nCurrent state (rows in columns):", current)
        display_board(current)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("Goal state reached!")
            return current

        neighbors = get_neighbors(current)

        print("\nNeighbor costs:")
        for neighbor in neighbors:
            print(neighbor, "->", calculate_cost(neighbor))

        best_cost = min(calculate_cost(n) for n in neighbors)

        if best_cost >= current_cost:
            print("Local minimum reached. Restarting...")
            return None

        best_neighbors = [
            n for n in neighbors
            if calculate_cost(n) == best_cost
        ]

        current = random.choice(best_neighbors)


def solve(initial):
    current = initial

    while True:
        result = hill_climbing(current)

        if result is not None:
            print("\nFinal solution:", result)
            print("Final cost:", calculate_cost(result))
            display_board(result)
            break

        current = [random.randrange(N) for _ in range(N)]
        print("\nNew initial state (rows in columns):", current)


initial = list(map(
    int, input("Enter current board (rows in columns, 0-3): ").split()
))

if len(initial) != N or any(x < 0 or x >= N for x in initial):
    print("Invalid input! Enter exactly 4 row values from 0 to 3.")
else:
    solve(initial)
```
