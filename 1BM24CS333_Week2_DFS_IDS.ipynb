```python
def get_neighbors(state):
    neighbors = []

    blank = state.index(0)
    row, col = divmod(blank, 3)

    moves = [(-1, 0), (1, 0),
             (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            board = list(state)

            board[blank], board[pos] = (
                board[pos], board[blank]
            )

            neighbors.append(tuple(board))

    return neighbors


# Depth-Limited DFS helper function
def depth_limited_dfs(state, goal, limit, path):
    if state == goal:
        return path

    if limit == 0:
        return None

    for next_state in get_neighbors(state):
        if next_state not in path:
            result = depth_limited_dfs(
                next_state,
                goal,
                limit - 1,
                path + [next_state]
            )

            if result is not None:
                return result

    return None


# DFS algorithm
def dfs(initial, goal, max_depth=20):
    print("\nRunning DFS...")

    return depth_limited_dfs(
        initial, goal, max_depth, [initial]
    )


# IDS algorithm
def ids(initial, goal, max_depth=30):
    print("\nRunning IDS...")

    for depth in range(max_depth + 1):
        print("Depth limit:", depth)

        solution = depth_limited_dfs(
            initial, goal, depth, [initial]
        )

        if solution is not None:
            return solution

    return None


# Print solution
def print_solution(solution):
    if solution is None:
        print("No solution found within the depth limit.")
        return

    print("\nSolution found!")
    print("Moves:", len(solution) - 1)

    for step, state in enumerate(solution):
        print("Step", step)

        for i in range(0, 9, 3):
            print(state[i:i + 3])

        print()


# Main program
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

print("8-Puzzle Problem")
print("1. Depth First Search (DFS)")
print("2. Iterative Deepening Search (IDS)")

choice = input("Enter your choice (1 or 2): ")

if choice not in ("1", "2"):
    print("Invalid choice!")

else:
    print("\nEnter the initial state (use 0 for blank):")
    print("Enter all 9 numbers separated by spaces.")

    try:
        initial = list(map(int, input().split()))

        # Allow input across three lines too
        while len(initial) < 9:
            initial.extend(map(int, input().split()))

        if len(initial) != 9 or set(initial) != set(range(9)):
            print("Invalid input! Enter numbers 0 to 8 exactly once.")

        else:
            initial = tuple(initial)

            if choice == "1":
                solution = dfs(initial, goal)
            else:
                solution = ids(initial, goal)

            print_solution(solution)

    except ValueError:
        print("Invalid input! Please enter integers only.")
```
